### 1. Authentication

Before we can place any orders, we must first authenticate our Kite Connect client. This gives us the necessary access to interact with our trading account.

```python
from kiteconnect import KiteConnect
import logging

logging.basicConfig(level=logging.INFO)

# --- Credentials ---
import os
API_KEY = os.environ.get('ZERODHA_ALGOTRADER_API_KEY')
API_SECRET = os.environ.get('ZERODHA_ALGOTRADER_API_SECRET')

access_token = None
kite = None

# --- Authentication ---
if api_key != "YOUR_API_KEY":
    try:
        kite = KiteConnect(api_key=api_key)
        print("Login URL:", kite.login_url())
        request_token = input("Enter request_token: ").strip()
        
        # Generate session and get access token
        data = kite.generate_session(request_token, api_secret=api_secret)
        access_token = data['access_token']
        
        # Set access token
        kite.set_access_token(access_token)
        
        # Verify authentication
        profile = kite.profile()
        logging.info(f"Logged in as: {profile['user_name']}")
        logging.info(f"Access Token: {access_token}")
        
    except Exception as e:
        logging.error(f"Authentication failed: {e}")
else:
    print("Please update your API_KEY and API_SECRET to run.")
```

### 2. Placing a BUY Order

We use the `kite.place_order()` method to send orders. Let's break down the parameters required to place a simple BUY market order for one share of Infosys (INFY) for delivery (CNC).

- `variety`: `kite.VARIETY_REGULAR`, `kite.VARIETY_AMO` (After Market Order), `kite.VARIETY_CO` (Cover Order), etc. The AMO and CO are covered in detail in later parts of the course.
- `exchange`: The exchange to place the order on (e.g., `kite.EXCHANGE_NSE`, `kite.EXCHANGE_NFO`).
- `tradingsymbol`: The symbol of the instrument (e.g., 'INFY').
- `transaction_type`: `kite.TRANSACTION_TYPE_BUY` or `kite.TRANSACTION_TYPE_SELL`.
- `quantity`: The number of units to trade.
- `product`: The product type:
  - `kite.PRODUCT_CNC`: For delivery-based equity trades.
  - `kite.PRODUCT_MIS`: For intraday trades (equities, F&O).
  - `kite.PRODUCT_NRML`: For overnight F&O positions.
- `order_type`: The type of order:
  - `kite.ORDER_TYPE_MARKET`: Executes at the current best available price.
  - `kite.ORDER_TYPE_LIMIT`: Executes at a specific price or better.
- `price`: The limit price for a `LIMIT` order.

To learn more about the order parameters, you can visit this link:
[Kite Order Parameters](https://kite.trade/docs/connect/v3/orders/)

```python
try:
    order_id = kite.place_order(
        variety=kite.VARIETY_REGULAR,             # Regular order
        exchange=kite.EXCHANGE_NSE,               # Place the order on NSE
        tradingsymbol='INFY',                     # The stock symbol

        transaction_type=kite.TRANSACTION_TYPE_BUY, # This is a BUY order
        
        quantity=1,                               # Number of shares to buy
        product=kite.PRODUCT_CNC,                 # For delivery-based trade (Cash and Carry)
        order_type=kite.ORDER_TYPE_MARKET         # This is a Market order
    )
    logging.info(f"BUY order placed successfully. Order ID: {order_id}")
except Exception as e:
    logging.error(f"Error placing BUY order: {e}")
```

### 3. Placing a SELL Order

Placing a SELL order is almost identical. The primary difference is changing the `transaction_type`.

Let's say we want to sell one share of Infosys that we hold in our portfolio.

```python
try:
    order_id = kite.place_order(
        variety=kite.VARIETY_REGULAR,
        exchange=kite.EXCHANGE_NSE,
        tradingsymbol='INFY',

        transaction_type=kite.TRANSACTION_TYPE_SELL, # This is a SELL order
        
        quantity=1,                                # Number of shares to sell
        product=kite.PRODUCT_CNC,
        order_type=kite.ORDER_TYPE_MARKET
    )
    logging.info(f"SELL order placed successfully. Order ID: {order_id}")
except Exception as e:
    logging.error(f"Error placing SELL order: {e}")
```
