# Cover Orders (CO)

Welcome! After learning how to place a separate stop-loss to protect a trade, it's natural to ask: can we do this all at once? The answer is yes.

In this notebook, we'll explore the **Cover Order (CO)**, a special order type for intraday traders that combines your entry and a compulsory stop-loss into a single, automated command.

**Disclaimer**: Cover Orders are designed specifically for intraday trading (`product='MIS'`) and may not be available for all stocks or segments.

### 1. Authentication

First, we need to authenticate our Kite Connect client to interact with our trading account.

```python
from kiteconnect import KiteConnect
import logging

logging.basicConfig(level=logging.INFO)

# --- Credentials & Initialisation ---
api_key = "YOUR_API_KEY"
access_token = "YOUR_ACCESS_TOKEN"

try:
    kite = KiteConnect(api_key=api_key)
    kite.set_access_token(access_token)
    profile = kite.profile()
    print(f"Logged in as {profile['user_name']}")
    logging.info("Authentication successful!")
except Exception as e:
    logging.error(f"Authentication failed: {e}")
```

### 2. Cover Orders (CO) - The 2-in-1 Deal

A Cover Order is a two-legged order that combines:
1.  Your initial entry order (which can be a Market or Limit order).
2.  A **compulsory** Stop-Loss Market (SL-M) order.

Think of it as mandatory insurance for your trade. It's a fantastic tool because it forces trading discipline; you cannot place a Cover Order without defining your maximum risk on the trade from the very beginning.

### 3. Placing a Cover Order

We're about to send a **Cover Order (CO)** to the exchange.

We've wrapped our code in a `try...except` block, which is just good practice. It means, "Try to place this order, but if something goes wrong, don't crash! Just tell me what the error was."


### Breaking Down the Order Details

Here’s a simple rundown of what we're telling the `kite.place_order` function to do:

* **`variety=kite.VARIETY_CO`**: We're explicitly telling the system, "This is a Cover Order."

* **`tradingsymbol="TATAMOTORS"`** & **`exchange=kite.EXCHANGE_NSE`**: We're placing an order for Tata Motors shares on the National Stock Exchange.

* **`quantity=10`**: We're looking to buy 10 shares.

* **`trigger_price=650`**: Here's our safety net. We're setting the stop-loss trigger at ₹650. If the stock price ever drops to this level, the system will automatically place an order to sell and limit our potential loss.

* **`order_type=kite.ORDER_TYPE_MARKET`** & **`transaction_type=kite.TRANSACTION_TYPE_BUY`**: This part is for our entry into the trade. We're telling the system to **BUY** our shares immediately at whatever the current **market price** is.

* **`product=kite.PRODUCT_MIS`**: We're flagging this as an intraday trade by setting the product to **MIS** (Margin Intraday Square off). This means we plan to close the position before the market shuts for the day.

### What Happens Next?

* **If it works**: The system will place the order, and we'll get a confirmation message with our unique `order_id`. Success!
* **If it fails**: If there's an issue (maybe not enough margin in the account or a bad trigger price), our `except` block will catch the error and print a helpful message telling us what went wrong.


```python
try:
    # --- Place the Cover Order ---
    order_id = kite.place_order(

        # Parameters for Cover Order (CO)
        variety=kite.VARIETY_CO,

        # Symbol and Exchange
        tradingsymbol="TATAMOTORS",
        exchange=kite.EXCHANGE_NSE,

        # Quantity and Stop-Loss
        quantity=10,
        trigger_price=650,  # This is the stop-loss trigger price

        # Buy Market Order
        order_type=kite.ORDER_TYPE_MARKET,
        transaction_type=kite.TRANSACTION_TYPE_BUY,

        # Product Type (Intraday)
        product=kite.PRODUCT_MIS,

    )
    logging.info(f"Cover Order placed successfully. Order ID: {order_id}")
except Exception as e:
    logging.error(f"Error placing CO: {e}")
```
